```python
!pip install transformers gradio sentencepiece -q
from transformers import AutoTokenizer, AutoModelForSeq2SeqLM
from transformers.utils import logging
import warnings
warnings.filterwarnings('ignore')
logging.set_verbosity_error()

MODEL_NAME = 'Helsinki-NLP/opus-mt-en-ur'

print('Loading model...')
tokenizer = AutoTokenizer.from_pretrained(MODEL_NAME)
model     = AutoModelForSeq2SeqLM.from_pretrained(MODEL_NAME)
print('Model ready!')
```

```python
def translate(text):
    if not text.strip():
        return "Please enter some English text!"

    inputs = tokenizer(
        text,
        return_tensors="pt",
        truncation=True,
        max_length=512
    )

    output = model.generate(
        **inputs,
        max_new_tokens=200,
        num_beams=4
    )

    translation = tokenizer.decode(
        output[0],
        skip_special_tokens=True
    )

    return translation


print("translate() function ready!")
```

```python
english_text = "I hope I am talking with my class teacher."

urdu_text = translate(english_text)

print("English:")
print(english_text)

print("\nUrdu:")
print(urdu_text)
```

```python
import gradio as gr

def translator_app(text):
    return translate(text)


app = gr.Interface(
    fn=translator_app,
    inputs=gr.Textbox(
        label="Enter English text",
        placeholder="Type your English sentence here to translate...",
        lines=6
    ),
    outputs=gr.Textbox(
        label="Urdu Translation",
        lines=6,
        rtl=True
    ),
    title="English to Urdu Translator",
    description="Translate English text into Urdu for free."
)

app.launch(share=True)
```
